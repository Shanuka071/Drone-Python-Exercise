```python
from pysimverse import Drone 
from pynput import keyboard
drone=Drone()   
drone.connect()
```

```python
print("--- MANUAL OVERRIDE ACTIVE ---")
print("W, A, S, D : Move Forward, Left, Back, Right")
print("UP / DOWN  : Change Altitude")
print("T / L      : Take-off / Land")
print("ESC        : Emergency Stop & Exit")

# 2. Define what happens when you push a key
def on_press(key):
    try:
        # Letters
        if key.char == 't': 
            drone.take_off()
        elif key.char == 'l': 
            drone.land()
        elif key.char == 'w': 
            drone.send_rc_control(0, 50, 0, 0)   # Forward
        elif key.char == 's': 
            drone.send_rc_control(0, -50, 0, 0)  # Backward
        elif key.char == 'a': 
            drone.send_rc_control(-50, 0, 0, 0)  # Left
        elif key.char == 'd': 
            drone.send_rc_control(50, 0, 0, 0)   # Right
            
    except AttributeError:
        # Special keys (Arrows)
        if key == keyboard.Key.up:
            drone.send_rc_control(0, 0, 50, 0)   # Up
        elif key == keyboard.Key.down:
            drone.send_rc_control(0, 0, -50, 0)  # Down
        elif key == keyboard.Key.esc:
            print("Stopping...")
            drone.send_rc_control(0, 0, 0, 0)
            drone.land()
            return False # This kills the listener

# 3. Define what happens when you LET GO of a key
def on_release(key):
    # This is the most important part: stop moving when key is released
    drone.send_rc_control(0, 0, 0, 0)
w
# 4. Start the 'Listener'
with keyboard.Listener(on_press=on_press, on_release=on_release) as listener:
    listener.join()
```
