```python
import pysimverse
import cv2 # This is OpenCV
import zmq # This is PyZMQ
import time

print("Environment Setup Successful!")
```

```python
from pysimverse import Drone
drone=Drone()
drone.connect()
```

```python
left_right = 0
forward_backward = 30  # Reduced from 50 for better control
up_down = 0
yaw_rotation = 0       # Yaw is usually a speed, not an angle, in RC control

# Fly forward for 3 seconds
start_time = time.time()
while time.time() - start_time < 3:
    # Send the movement command
    drone.send_rc_control(left_right, forward_backward, up_down, yaw_rotation)
    time.sleep(0.05)  # Crucial: This prevents the drone from "getting mad"

# STOP the drone (Reset all values to 0)
drone.send_rc_control(0, 0, 0, 0)
print("Movement finished, drone stopped.")

drone.land()
```

```python
import time
from pysimverse import DroneSim

drone = DroneSim()
drone.take_off()

# Define your movement variables
left_right = 0
forward_backward = 50
up_down = 0
yaw = 0

# 1. SET A DURATION (e.g., 3 seconds)
start_time = time.time()

# 2. CHANGE 'True' TO A CONDITION
# This loop will run for exactly 3 seconds
while time.time() - start_time < 3:
    drone.send_rc_control(left_right, forward_backward, up_down, yaw)
    time.sleep(0.05)

# 3. SAFETY STOP
# Send 0s to stop the drone's momentum before landing
print("Stopping movement...")
drone.send_rc_control(0, 0, 0, 0)
time.sleep(1) # Let it hover for a second

# 4. LAND
print("Landing...")
drone.land()
```
